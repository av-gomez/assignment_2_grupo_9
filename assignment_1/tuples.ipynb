## 1. Crear la tupla y mostrarla

````python
ages = (20, 25, 22, 30, 28)

ages
````

Defino la tupla con paréntesis, que es lo que la distingue de una lista, y la
muestro escribiendo `ages` en la última línea.

## 2. Primer y último elemento

````python
extremos = {
    "Primer elemento": ages[0],
    "Último elemento": ages[-1],
}

for etiqueta, valor in extremos.items():
    print(f"{etiqueta}: {valor}")
````

Uso el índice `0` para el primero y `-1` para el último, que me evita tener que
saber cuántos elementos hay. Los junto en un diccionario para mostrarlos con un
solo `print()`.

## 3. Valor máximo, mínimo y número de elementos

````python
resumen = {
    "Valor máximo": max(ages),
    "Valor mínimo": min(ages),
    "Número de elementos": len(ages),
}

for etiqueta, valor in resumen.items():
    print(f"{etiqueta}: {valor}")
````

`max()`, `min()` y `len()` funcionan igual que en una lista porque solo leen la
tupla. Lo que no se puede es modificarla, que es lo que pruebo abajo.

Dejo la línea comentada para que el notebook se ejecute completo, como permite
el enunciado. Si la descomento, Python lanza:

```
TypeError: 'tuple' object does not support item assignment
```

**Por qué no lo permite:** las tuplas son inmutables, es decir, el tipo `tuple` no
implementa la asignación por índice (`__setitem__`). Esto sirve para proteger datos
que no deben cambiar y permite usar tuplas como claves de un diccionario, cosa que
con listas no se puede. Si necesito otro valor, creo una tupla nueva:
`ages = (21,) + ages[1:]`.
