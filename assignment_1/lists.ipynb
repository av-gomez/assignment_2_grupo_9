## 1. Crear la lista y mostrarla

```python
grades = [15, 18, 12, 20, 16]

grades
```

Creo la lista con las cinco notas y escribo `grades` en la última línea para
mostrarla. No uso `print()` porque Jupyter ya muestra el valor de la última
línea de la celda.

## 2. Agregar la nota 17

```python
grades.append(17)

grades
```

`append()` agrega el 17 al final y modifica la lista original. Por eso lo llamo
solo, sin asignarlo: `append()` devuelve `None`, así que `grades = grades.append(17)`
dejaría la lista vacía.

## 3. Ordenar de menor a mayor

```python
grades.sort()

grades
```

Uso `sort()` y no `sorted()` porque el enunciado pide ordenar la lista, no crear
una copia: `sort()` reordena la misma lista, `sorted()` devolvería una nueva.

## 4. Nota máxima, nota mínima y total de notas

```python
resumen = {
    "Nota más alta": max(grades),
    "Nota más baja": min(grades),
    "Total de notas": len(grades),
}

for etiqueta, valor in resumen.items():
    print(f"{etiqueta}: {valor}")
```

Guardo los tres resultados en un diccionario y los recorro con un bucle. Lo hice
así para usar un solo `print()` en lugar de tres líneas casi idénticas.

## 5. Lista final

```python
grades
```

Muestro el resultado final: la lista ya tiene el 17 agregado y está ordenada de
